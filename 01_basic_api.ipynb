# 2. 创建张量的方法

```python
# 1. 使用列表创建Tensor
t1 = torch.Tensor([1, 2, 3])
print(t1)
"""
tensor([1., 2., 3])
"""

# 2. 使用numpy数组创建Tensor
array1 = np.arange(12).reshape(3, 4)

t2 = torch.Tensor(array1)
print(t2)
```

```python
# 3. 通过torch的API创建Tensor
"""
torch.empty(3, 4)：创建3行4列的空的tensor，会用无用的数据进行填充
torch.ones([3, 4])：3行4列全为1的tensor
torch.zeros([3, 4])：3行4列全为0的tensor
torch.rand([3, 4])：3行4列随机值在[0, 1]之间的值
torch.randint(low = 0, high = 10, size = [3, 4]) 创建3 * 4的随机整数Tensor，值区间：[low, high]
torch.randn([3, 4])均值为0，方差为1，3*4 的tensor
"""
print(torch.empty(3, 4))
print(torch.ones([3, 4]))
print(torch.zeros([3, 4]))
# 每个元素从[0, 1)的均匀分布里随机采样(包含0，不包含1)
print(torch.rand([3, 4]))
```

```python
# 每个元素从[0, 10)里随机取（包含0，不包含10），也就是取值只能是 0～9
print(torch.randint(low = 0, high = 10, size = [3, 4]))
"""
标准化 != 正态化，但正态化一定是标准化
randn = random + normal
    由于正态分布曲线与x轴面积的问题，68%的数值落在 (-1, 1) 之间
"""
print(torch.randn([3, 4]))
```

# 3. 张量的方法和属性

```python
# 1. tensor.item()，当tensor中只有一个元素可以用的时候
a = torch.tensor(np.arange(1))
print(a)
# 只有一个元素的时候可以用，返回一个python的标量，不可以用于多个元素的tensor
print(a.item())
print('--' * 50)
# .item()只要求张量里元素总数为1，跟它的形状/维度无关；也就是必须只有一个元素，而不是“必须是0维”
print(torch.Tensor([[[1]]]).item())
```

```python
# 2. Tensor转为ndarray
t2 = torch.Tensor([3, 4])
print(t2.numpy())

print(t2.shape)
print(t2.size())
print(t2.size(0))
```

```python
# 2. Tensor转为ndarray
"""
t2 = torch.Tensor([[3, 4]]) -> 创建一个 1*2 的张量：
    用torch.Tensor（大写T）默认是float32，所以3、4变成 3. 、4.
t2.numpy() -> 把张量转成 Numpy 数组：
    前提：张量在CPU上且requires_grad=False，否则会报错
t2.shape:
    第0维大小 = 1（1行）
    第1维大小 = 2（2列）
t2.size():
    size()不传参时，等价于shape
    注：shape是属性，size()是方法，结果一样
t2.size(1):
    返回第1维（从0开始数）的大小
    t2.size(0) -> 1
    t2.size(1) -> 2
"""
t2 = torch.Tensor([[3, 4]])
print(t2.numpy())
# 两种获取形状的方式
print(t2.shape)
print(t2.size())
# 返回第一维的大小
print(t2.size(1))
```

```python
t2
```

```python
# 写一个ndarray
"""
为什么id不同但内存相同
    id()            Python对象标识，不同对象就不同，array1和array2是两个不同的Python对象
    底层数据         可以多个对象共享同一块内存，reshape通常不复制数据
    reshape创建了一个新的ndarray对象（新的shape、新的strides），但它指向原数组的数据缓冲区
如果无法用“改变视图”的方式完成reshape（比如内存不连续），NumPy会复制：
    array1 = np.array([[1, 2, 3], [4, 5, 6]])
    array2 = array1.T.reshape(3, 2)   # 转置后不连续，可能复制
    print(np.shares_memory(array1, array2))   # 可能 False
        此时array2和array1完全独立
"""
array1 = np.array([[1, 2, 3], [4, 5, 6]])
print(id(array1))
array2 = array1.reshape(3, 2)
print(id(array2))
```

```python
# 此时修改array2时，array1也会同时改变
array2[0, 0] = 100
print(array1)
print(array2)
```

```python
print(array1.ndim)
print(array2.ndim)
```

```python
# 3. 形状修改，tensor.view((3, 4))，类似numpy中的reshape，是一种浅拷贝，仅仅形状发生改变，返回一个新的结果
t2 = torch.Tensor([[[3, 4]]])
print(t2.size())                                    # 三维张量
print(t2.view([1, 2]))                              # 二维张量（1行2列）
print(t2.view([2]))                                 # 一维张量（2列）
b = t2.view([2, -1])                                # -1表示自动计算
print(b)
print('--' * 50)
print(t2)                                           # t2的形状并没有发生改变
# https://pytorch.org/docs/stable/tensor_view.html
t2.storage().untyped().data_ptr() == b.storage().untyped().data_ptr()   # 判断两个tensor是否共享内存
```

```python
b[0, 0] = 100
print(b)
print(t2)
```

```python
"""
torch.Tensor是一个类（构造函数），torch.tensor是一个工厂函数。前者行为多、易踩坑，后者更推荐用于日常创建张量。
torch.Tensor：
    默认float32
    torch.Tensor是torch.FloatTensor的别名。永远创建float32，即使传入整数也会被转成浮点。
        # torch.Tensor 把参数当"形状"
        a = torch.Tensor(2, 3)
        print(a.shape)   # torch.Size([2, 3])   创建 2×3 的未初始化张量
        
        torch.Tensor([2, 3])     # 创建一维张量，值为 [2, 3]
torch.tensor:
    会根据输入数据自动推断dtype，这是它更符合直觉的地方。
        # torch.tensor 把参数当"数据"
        b = torch.tensor(2, 3)   # 报错！tensor() 只接受一个数据参数
        
        torch.tensor([2, 3])     # 与上面torch.Tensor([2, 3])相同，同样是一维张量，值为 [2, 3]

区分：
    torch.Tensor(*sizes)可以接收多个整数作为形状；torch.tensor(data)只接收一个数据对象。
"""
t4 = torch.tensor(np.arange(24).reshape(2, 3, 4))
print(t4.shape)                                                     # torch.Size([2, 3, 4])
print('--' * 50)
print(t4.transpose(0, 1).shape)                                     # 交换0轴和1轴 torch.Size([3, 2, 4])
print('--' * 50)
print(t4.permute(1, 0, 2).shape)                                    # shape = 3 * 2 * 4 torch.Size([3, 2, 4])
print('--' * 50)
print(t4.permute(1, 2, 0).shape)                                    # shape = 3 * 4 * 2 torch.Size([3, 4, 2])
print('--' * 50)
print(t4.permute(2, 1, 0).shape)                                    # shape = 4 * 3 * 2 torch.Size([4, 3, 2])
```

```python
# 在方法后加_，会原地修改
x = torch.tensor(np.arange(12).reshape(3, 4), dtype=torch.int8)
print(x)
y = torch.ones([3, 4], dtype=torch.int64)
print(y)
print('--' * 50)

x.sub_(y)                               # sub_ 原地修改，不加下划线就会创建一个新的tensor来存储
print(x)
```

```python
#初始化两个张量，一个3,4，一个3,1,运算与ndarray相同
t8 = torch.tensor(np.arange(12).reshape(3,4),dtype=torch.float32)
t9 = torch.tensor(np.arange(3).reshape(3,1),dtype=torch.float32)
print(t8)
print(t9)
print(t8-t9)
```

```python
"""
GPU中tensor的使用：
1.实例化device， torch.device("cpu" or "cuda:0"),选cuda:0需要torch.cuda.is_available()==true
2.tensor.to(device),把tensor转化为CUDA支持的tensor或者CPU支持的tensor
"""
print(torch.cuda.is_available())
if torch.cuda.is_available():
    device = torch.device("cuda") #cuda device对象
    """
    创建一个和x形状、数据类型都相同的全1张量，并把它放到指定设备device上
    torch.ones_like                 生成一个和输入张量同形状的全1张量
    x                               模版张量，决定y的形状和dtype
    device=device                   指定张量放在那个设备（cpu/cuda/mps）
    y                               结果张量
    """
    y = torch.ones_like(x, device = device) #创建一个在cuda熵的tensor
    x = x.to(device) #把x转为cuda上的tensor
    z = x+y
    print(z)
    print(z.to("cpu", torch.double))

# torch.tensor([1.9806], device = "cuda:0")
device = torch.device("cpu")
x.to(device)
```

```python
# 张量的运算
t8 = torch.tensor(np.arange(12).reshape(3,4),dtype=torch.float32)
print(t8)
print('--' * 50)
t9 = torch.tensor(np.arange(12, 24).reshape(3,4),dtype=torch.float32)
print(t9)
print('--' * 50)
t8*t9
```

```python
"""
transpose返回的是视图，不复制数据
mm：matrix multiplication（矩阵相乘）
torch.mm                                矩阵乘法（只支持二维张量）
t8                                      左矩阵
t9.transpose(0, 1)                      把t9的第0维和第1维交换（即转置）
"""
torch.mm(t8, t9.transpose(0,1)) #矩阵乘法
```
