```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

x = np.linspace(-4, 4, 1000)
y = norm.pdf(x, 0, 1)   # 标准正态密度

plt.figure(figsize=(8, 5))
plt.plot(x, y, 'b-', linewidth=2, label='N(0, 1)')

# 填充三个区间
plt.fill_between(x, y, where=(x >= -1) & (x <= 1),
                 color='red', alpha=0.3, label='68.27% (-1 ~ 1)')
plt.fill_between(x, y, where=(x >= -2) & (x <= 2),
                 color='green', alpha=0.15, label='95.45% (-2 ~ 2)')

# 标注
plt.axvline(0, color='gray', linestyle='--', alpha=0.5)
plt.title('Standard Normal Distribution N(0,1)')
plt.xlabel('x')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```
